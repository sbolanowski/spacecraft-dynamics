# Euler Parameters (Quaternions)

Euler parameters, commonly known as quaternions, are a mathematical representation of orientation and rotation in three-dimensional space. They are an extension of complex numbers and provide a compact and efficient way to represent rotations without some of the problems associated with other representations like Euler angles (e.g., gimbal lock). Quaternions are widely used in computer graphics, robotics, aerospace engineering, and other fields that require 3D rotation computations.

## Definition

A quaternion is typically represented as:

$$
q = b_0 + b_1 \hat{i} + b_2 \hat{j} + b_3 \hat{k}
$$

Where:
- b₀ is the scalar (real) part.
- b₁, b₂, b₃ are the components of the vector (imaginary) part.
- î, ĵ, k̂ are the unit vectors of the quaternion corresponding to the x, y, and z axes, respectively.

The quaternion can also be written as:

$$
q = [b_0, b_1, b_2, b_3]
$$

## Relationship to Rotation

Quaternions can represent rotations by encoding the rotation axis and the angle. A unit quaternion that represents a rotation by an angle θ about a unit vector v̂ = [v₁, v₂, v₃] is given by:

$$
q = \cos\left(\frac{\theta}{2}\right) + \hat{v} \sin\left(\frac{\theta}{2}\right)
$$

Where:
- θ is the angle of rotation (in radians).
- v̂ = [v₁, v₂, v₃] is the unit vector representing the axis of rotation.


## Example: Rotating a Vector Using Quaternions

Given a quaternion q = [q₀, q₁, q₂, q₃], we can rotate a vector v = [vₓ, vᵧ, v𝓏] using quaternion multiplication. The quaternion multiplication steps are:

1. Convert the vector v into a pure quaternion: qᵥ = [0, vₓ, vᵧ, v𝓏].
2. Perform the rotation as:


$$
q' = q \cdot q_v \cdot q^{-1}
$$

Where q⁻¹ is the inverse of the quaternion q.

The resulting quaternion q' will contain the rotated vector in its vector part.

## Advantages of Using Quaternions

- **No Gimbal Lock**: Unlike Euler angles, quaternions do not suffer from gimbal lock, which can occur when one of the axes aligns and the system loses a degree of freedom.
- **Compact**: Quaternions are more compact (4 parameters) compared to rotation matrices (9 parameters).
- **Efficient Computation**: Quaternion operations are computationally more efficient than matrix multiplications or Euler angle conversions.
